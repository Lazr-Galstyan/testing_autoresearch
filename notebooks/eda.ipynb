# E-Commerce Churn — Exploratory Data Analysis

Goal: understand the dataset before feeding it into the traditional and autoresearch pipelines.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

RANDOM_SEED = 42
DATA_DIR = Path('..') / 'data' / 'raw'
```

## 1. Load Data

```python
# Update filename once the Excel file is placed in data/raw/
df = pd.read_excel(DATA_DIR / 'churn_data.xlsx')
print(df.shape)
df.head()
```

## 2. Schema & Data Types

```python
df.info()
```

## 3. Missing Values

```python
missing = df.isnull().sum()
missing[missing > 0].sort_values(ascending=False)
```

## 4. Target Distribution

```python
target_col = 'Churn'  # update if column name differs
ax = df[target_col].value_counts().plot(kind='bar')
ax.set_title('Churn Distribution')
plt.tight_layout()
plt.show()
print(df[target_col].value_counts(normalize=True).round(3))
```

## 5. Numeric Feature Distributions

```python
num_cols = df.select_dtypes(include=np.number).columns.tolist()
df[num_cols].hist(figsize=(14, 10), bins=30)
plt.tight_layout()
plt.show()
```

## 6. Correlation Heatmap

```python
plt.figure(figsize=(12, 8))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

## 7. Categorical Features vs Churn

```python
cat_cols = df.select_dtypes(include='object').columns.tolist()
for col in cat_cols:
    ct = pd.crosstab(df[col], df[target_col], normalize='index').round(3)
    print(f'\n--- {col} ---')
    print(ct)
```
